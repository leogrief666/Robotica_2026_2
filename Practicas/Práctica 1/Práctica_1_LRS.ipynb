# Previo de la Práctica 1

1. ¿Qué es el modelado con Xacro? 

    Es una herramienta de estensión para los archivos URDF en ROS. Mientras que un archivo URDF estándar es un XML estático (esto nos hace tener que escribir líneas repetitivas para cada eslabón o articulación). También nos permite utilizar una estructura de programación modular.

Sus pruncipales ventajas son:

   - Reducción del código: Nos permite definir constantes y reutilizarlas en todo el modelo.
   - Uso de macros: Se pueden crear funciones para componentes que se repiten.
   - Expresiones matemáticas: Nos permite realizar cálculos dentro de un archivo.

En resumen, un Xacro es como dejar de escribir un texto a mano para empezar a usar una plantilla inteligente; nos permite que el modelo del robot sea más fácil de mantener, leer y modificar.

2. Definición de joint y eslabón.

    Son los dos bloques contructivos que usa cualquier robot en ROS:

   - Eslabón: Es el componente físico rígido de nuestro robot. Un _link_ posee propiedades físicas como:
      - Geometría: Su forma visual.
      - Inercia: Su masa y la distribución de ésta misma.
      - Colisión: El modelo simplificado que usa el simulador para detectar si el robot choca con algo.
</p>

   - Articulación: Ésta es la conexión o unión entre dos links. El _joint_ define cómo es que se mueven los eslabones entre sí. Esto también depende del tipo de _joint_, que determina la libertad de movimiento:
    
      - Revoluta: Permite giros (como la articulación de un codo o la vuelta de una llanta).
      - Prismática: Permite un desplazamiento lineal.
      - Fija: Une dos eslabones sin permitie el movimiento de los mismos.

2. Configuración del launch en python:

    import os
    from ament_index_python.packages import get_package_share_directory
    from launch import LaunchDescription
    from launch.actions import ExecuteProcess
    from launch_ros.actions import Node
    from launch.substitutions import Command

    def generate_launch_description():

    pkg_scara_description = get_package_share_directory('scara_description')
    pkg_scara_bringup = get_package_share_directory('scara_bringup')

    urdf_path = os.path.join(pkg_scara_description, 'urdf', 'gz2_scara.xacro')
    gazebo_config_path = os.path.join(pkg_scara_bringup, 'config', 'gz_bridge.yaml')
    world_path = os.path.join(pkg_scara_bringup, 'worlds', 'world_config.world')

    rviz_config_path = os.path.join(pkg_scara_bringup, 'rviz', 'scara_config.rviz')

    use_sim_time = {'use_sim_time': True}

    robot_state_publisher_node = Node(
        package='robot_state_publisher',
        executable='robot_state_publisher',
        parameters=[{'robot_description': Command(['xacro ', urdf_path])}, use_sim_time]
    )


    gz_ros_bridge_node = Node(
        package='ros_gz_bridge',
        executable='parameter_bridge',
        parameters=[{'config_file': gazebo_config_path}, use_sim_time],
        output='screen'
    )

    gz_sim = ExecuteProcess(
        cmd=['gz', 'sim', '-r', world_path, '--render-engine', 'ogre'],
        output='screen'
    )

    spaw_entity = Node(
    package='ros_gz_sim',
    executable='create',
    arguments=[
        '-topic', 'robot_description',
        '-name', 'robot_scara',
        '-world', 'world_config' # DEBE coincidir con el <world name="..."> de tu archivo .world
    ],
    output='screen'
    )

    rviz_node = Node(
        package='rviz2',
        executable='rviz2',
        name='rviz2',
        output='screen',
        arguments=['-d', rviz_config_path],
        parameters=[use_sim_time] # Fundamental para evitar errores de TF
    )

    return LaunchDescription([
        robot_state_publisher_node,
        gz_sim,
        spaw_entity,
        gz_ros_bridge_node,
        rviz_node
    ])
